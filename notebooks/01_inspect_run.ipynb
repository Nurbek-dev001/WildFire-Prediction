```python
index = pd.read_csv(RUN / 'map' / 'map_index.csv')
with rasterio.open(RUN / 'map' / index.prob_file.iloc[-1]) as r:
    arr = r.read(1, masked=True)
    print(r.tags()['TARGET_DATE_UTC'], r.crs)
plt.imshow(arr, cmap='Blues'); plt.colorbar(label='calibrated probability (hazard, not risk)');
```
